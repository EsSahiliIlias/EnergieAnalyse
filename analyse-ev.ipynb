```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df_ev_consumption = pd.read_excel('data/Consumption of Orac 01-05-2025 tot 01-05-2026.xlsx')

df_ev_consumption.head()
```

```python
df_ev_consumption.dtypes
```

```python
#tel all columns that are of type number and make a colum 'total_consumption_ev' that is the sum of all those columns
numeric_cols = df_ev_consumption.select_dtypes(include=[np.number]).columns
df_ev_consumption['total_consumption_ev'] = df_ev_consumption[numeric_cols].sum(axis=1)
df_ev_consumption.head()
```

```python
#delete all columns expect 'total_consumption_ev' & 'date'
df_ev_consumption = df_ev_consumption[['Date', 'total_consumption_ev']]
df_ev_consumption.head()
```

```python
#zet date naar datetime
df_ev_consumption['Date'] = pd.to_datetime(df_ev_consumption['Date'])
df_ev_consumption.dtypes
```

```python
# === SEIZOEN ANALYSE: ALLE SEIZOENEN (2025 & 2026) ===

# Voorbereiding EV data
dfe_extended = df_ev_consumption.copy()

# Zorg datetime index
if not pd.api.types.is_datetime64_any_dtype(dfe_extended.index):
    possible_time_cols = [c for c in dfe_extended.columns if 'date' in c.lower() or 'datum' in c.lower()]
    if len(possible_time_cols) > 0:
        timecol = possible_time_cols[0]
    else:
        timecol = dfe_extended.columns[0]
    dfe_extended[timecol] = pd.to_datetime(dfe_extended[timecol], errors='coerce')
    dfe_extended = dfe_extended.sort_values(timecol).set_index(timecol)

# Bereken totaal EV per timestamp
ev_total_extended = dfe_extended.sum(axis=1)

# Definieer seizoenen
seasons_extended = {
    'Zomer 2025\n(Jun - Aug 2025)': [
        (2025, 6),   # Jun
        (2025, 7),   # Jul
        (2025, 8)    # Aug
    ],
    'Herfst 2025\n(Sep - Nov 2025)': [
        (2025, 9),   # Sep
        (2025, 10),  # Okt
        (2025, 11)   # Nov
    ],
    'Winter 2025/26\n(Dec 2025 - Feb 2026)': [
        (2025, 12),  # Dec 2025
        (2026, 1),   # Jan 2026
        (2026, 2)    # Feb 2026
    ],
    'Lente 2026\n(Mrt - Mei 2026)': [
        (2026, 3),   # Mrt
        (2026, 4),   # Apr
        (2026, 5)    # Mei
    ]
}

# Plot 6 grafieken (3x2)
fig, axes = plt.subplots(2, 3, figsize=(18, 10))
fig.suptitle('EV Verbruik per Seizoen - Volledige Periode (Zomer 2025 - Mei 2026)', fontsize=16, fontweight='bold')
axes = axes.flatten()

season_stats_ext = {}
colors_extended = ['#FF8C00', '#DC143C', '#4169E1', '#90EE90']  # oranje, rood, blauw, groen

for idx, (season_name, months) in enumerate(seasons_extended.items()):
    # Filter data voor deze seizoen
    mask = pd.Series([False] * len(ev_total_extended), index=ev_total_extended.index)
    for year, month in months:
        mask |= ((ev_total_extended.index.year == year) & (ev_total_extended.index.month == month))
    
    season_data = ev_total_extended[mask]
    
    if len(season_data) == 0:
        print(f"Geen data gevonden voor {season_name}")
        continue
    
    ax = axes[idx]
    
    # Plot
    ax.plot(season_data.index, season_data.values, 
            linewidth=2.5, color=colors_extended[idx], label=season_name, marker='o', markersize=3, alpha=0.7)
    
    ax.set_xlabel('Datum', fontsize=11, fontweight='bold')
    ax.set_ylabel('Verbruik [kWh/uur]', fontsize=11, fontweight='bold')
    ax.set_title(season_name, fontsize=12, fontweight='bold')
    
    ax.grid(True, alpha=0.3)
    ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda x, p: f'{x:,.0f}'))
    
    # Statistieken
    avg = season_data.mean()
    peak = season_data.max()
    total = season_data.sum()
    min_val = season_data.min()
    
    season_stats_ext[season_name] = {
        'avg': avg,
        'peak': peak,
        'min': min_val,
        'total': total,
        'records': len(season_data)
    }
    
    # Annotatie op grafiek
    textstr = f'Gem: {avg:.1f} kWh/uur\nPiek: {peak:.1f} kWh/uur\nMin: {min_val:.1f} kWh/uur'
    ax.text(0.02, 0.98, textstr, transform=ax.transAxes, 
            fontsize=10, verticalalignment='top',
            bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.8))

# 5e grafiek (idx=4): Totalen vergelijking
ax = axes[4]
season_names_list = list(seasons_extended.keys())
totals = [season_stats_ext[s]['total'] for s in season_names_list if s in season_stats_ext]
season_labels_short = [s.split('\n')[0] for s in season_names_list if s in season_stats_ext]

x_pos = range(len(totals))
bars_tot = ax.bar(x_pos, totals, color=colors_extended[:len(totals)], alpha=0.8, edgecolor='black', linewidth=1.5)

ax.set_ylabel('Totaal per seizoen [kWh]', fontsize=11, fontweight='bold')
ax.set_title('Totaal Verbruik per Seizoen', fontsize=12, fontweight='bold')
ax.set_xticks(x_pos)
ax.set_xticklabels(season_labels_short, fontsize=10, rotation=45)
ax.grid(True, alpha=0.3, axis='y')

# Add value labels on bars
for bar in bars_tot:
    height = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2., height,
            f'{height:,.0f}', ha='center', va='bottom', fontsize=9, fontweight='bold')

# 6e grafiek (idx=5): Gemiddelde en Piek vergelijking
ax = axes[5]
avgs = [season_stats_ext[s]['avg'] for s in season_names_list if s in season_stats_ext]
peaks = [season_stats_ext[s]['peak'] for s in season_names_list if s in season_stats_ext]

x_pos2 = range(len(avgs))
width = 0.35

bars_avg = ax.bar([i - width/2 for i in x_pos2], avgs, width, label='Gemiddeld (kWh/uur)', color=colors_extended[:len(avgs)], alpha=0.7)
bars_peak = ax.bar([i + width/2 for i in x_pos2], peaks, width, label='Piek (kWh/uur)', color='darkred', alpha=0.6)

ax.set_ylabel('kWh/uur', fontsize=11, fontweight='bold')
ax.set_title('Gemiddelde vs Piek per Seizoen', fontsize=12, fontweight='bold')
ax.set_xticks(x_pos2)
ax.set_xticklabels(season_labels_short, fontsize=10, rotation=45)
ax.legend(fontsize=10)
ax.grid(True, alpha=0.3, axis='y')

plt.tight_layout()
plt.show()

# Print statistieken
print('\n' + '='*70)
print('📊 SEIZOEN ANALYSE: VOLLEDIGE PERIODE (ZOMER 2025 - MEI 2026)')
print('='*70)

for season_name, stats in season_stats_ext.items():
    print(f"\n{season_name.replace(chr(10), ' ')}:")
    print(f"  Records: {stats['records']}")
    print(f"  Gemiddeld: {stats['avg']:.2f} kWh/uur")
    print(f"  Piek: {stats['peak']:.2f} kWh/uur")
    print(f"  Minimum: {stats['min']:.2f} kWh/uur")
    print(f"  Totaal: {stats['total']:,.0f} kWh")

print('\n' + '='*70)
```

```python
df_verbruik_september = pd.read_excel('data/electricity verbruik totaal alles - 2025-09-01 - 2025-09-30 - 15min.xlsx')

df_verbruik_september.head()
```

```python
#Set datum als index
df_verbruik_september['Datum'] = pd.to_datetime(df_verbruik_september['Datum'])
df_verbruik_september.set_index('Datum', inplace=True)

df_ev_consumption['Date'] = pd.to_datetime(df_ev_consumption['Date'])
df_ev_consumption.set_index('Date', inplace=True)
```

```python
# === SEPTEMBER ANALYSE: EV INVLOED OP TOTAAL VERBRUIK ===

# Filter EV-data voor september
# df_ev_consumption bevat hourly kWh per connector
df_september_ev = df_ev_consumption[(df_ev_consumption.index.month == 9)].copy()

# Voorbereiding facility verbruik data
df_verbruik_sep = df_verbruik_september.copy()

# Zet eerste kolom naar datetime (indien nodig) en index
if df_verbruik_sep.index.dtype == 'O':
    timestamp_col_fac = df_verbruik_sep.columns[0]
    df_verbruik_sep[timestamp_col_fac] = pd.to_datetime(df_verbruik_sep[timestamp_col_fac], errors='coerce')
    df_verbruik_sep = df_verbruik_sep.sort_values(timestamp_col_fac)
    df_verbruik_sep.set_index(timestamp_col_fac, inplace=True)

print("=== SEPTEMBER DATA PREP ===")
print(f"EV (hourly) records: {len(df_september_ev)}")
print(f"Facility (15min) records: {len(df_verbruik_sep)}")

# Bereken totale EV per uur in september (kWh per uur)
ev_total_sep = df_september_ev.sum(axis=1)  # already hourly kWh

# Facility: kolom 'Verbruik [kW]' is vermogen per 15min (kW)
# Converteer naar energie (kWh) per 15min door *0.25, en som naar kWh per uur
if 'Verbruik [kW]' in df_verbruik_sep.columns:
    facility_kwh_15min = df_verbruik_sep['Verbruik [kW]'] * 0.25  # kWh per 15min
else:
    # fallback: pak eerste numerieke kolom
    num_cols = df_verbruik_sep.select_dtypes(include=[np.number]).columns
    if len(num_cols) == 0:
        raise ValueError('Geen numerieke kolommen gevonden in facility dataset')
    facility_kwh_15min = df_verbruik_sep[num_cols[0]] * 0.25

# Resample 15min -> hourly kWh
facility_kwh_hour = facility_kwh_15min.resample('H').sum()

# Align indexes to common hourly index (intersection)
common_index = ev_total_sep.index.intersection(facility_kwh_hour.index)
ev_aligned = ev_total_sep.loc[common_index]
fac_aligned = facility_kwh_hour.loc[common_index]

# === VISUALISATIE ===
fig, axes = plt.subplots(3, 1, figsize=(16, 12))

# Plot 1: Beide dataseries (kWh per uur)
ax = axes[0]
ax2 = ax.twinx()

line1 = ax.plot(ev_aligned.index, ev_aligned.values, 
                linewidth=2.5, color='steelblue', label='EV Verbruik (kWh/uur)', alpha=0.9)
ax.set_ylabel('EV Verbruik [kWh/uur]', fontsize=11, fontweight='bold', color='steelblue')
ax.tick_params(axis='y', labelcolor='steelblue')

line2 = ax2.plot(fac_aligned.index, fac_aligned.values, 
                 linewidth=2.5, color='crimson', label='Facility Verbruik (kWh/uur)', alpha=0.6)
ax2.set_ylabel('Facility Verbruik [kWh/uur]', fontsize=11, fontweight='bold', color='crimson')
ax2.tick_params(axis='y', labelcolor='crimson')

ax.set_title('September 2025: EV vs Facility Verbruik (kWh/uur)', fontsize=13, fontweight='bold')
ax.grid(True, alpha=0.3)

lines = line1 + line2
labels = [l.get_label() for l in lines]
ax.legend(lines, labels, loc='upper left', fontsize=11)

# Plot 2: Dagelijkse totalen (kWh/dag)
ax = axes[1]
ev_daily = ev_aligned.resample('D').sum()
fac_daily = fac_aligned.resample('D').sum()

x_pos = range(len(ev_daily))
width = 0.35

bars1 = ax.bar([i - width/2 for i in x_pos], ev_daily.values, width, label='EV Verbruik (kWh/dag)', color='steelblue', alpha=0.8)
bars2 = ax.bar([i + width/2 for i in x_pos], fac_daily.values, width, label='Facility Verbruik (kWh/dag)', color='crimson', alpha=0.8)

ax.set_xlabel('Dag van september', fontsize=11, fontweight='bold')
ax.set_ylabel('kWh/dag', fontsize=11, fontweight='bold')
ax.set_title('Dagelijkse Verbruik Vergelijking (kWh/dag)', fontsize=13, fontweight='bold')
ax.set_xticks(x_pos)
ax.set_xticklabels([f'{i+1}' for i in range(len(ev_daily))], fontsize=9)
ax.legend(fontsize=11)
ax.grid(True, alpha=0.3, axis='y')

# Plot 3: Percentage EV van totaal per uur
ax = axes[2]

ev_pct = (ev_aligned / (ev_aligned + fac_aligned) * 100).fillna(0)

ax.plot(ev_pct.index, ev_pct.values, linewidth=2.5, color='darkgreen', marker='o', markersize=3, alpha=0.7)
ax.fill_between(ev_pct.index, ev_pct.values, alpha=0.2, color='darkgreen')
ax.set_ylabel('EV Aandeel [%]', fontsize=11, fontweight='bold')
ax.set_title('EV als Percentage van Totaal Verbruik per uur', fontsize=13, fontweight='bold')
ax.set_ylim(0, 100)
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# === STATISTIEKEN ===
print("\n" + "="*70)
print("📊 SEPTEMBER ANALYSE: EV INVLOED (gereviseerd)")
print("="*70)

print(f"\n🔌 EV VERBRUIK (kWh):")
print(f"  Totaal: {ev_aligned.sum():,.0f} kWh")
print(f"  Gemiddeld per uur: {ev_aligned.mean():.2f} kWh/uur")
print(f"  Piek: {ev_aligned.max():.2f} kWh/uur op {ev_aligned.idxmax()}")
print(f"  Minimum: {ev_aligned.min():.2f} kWh/uur op {ev_aligned.idxmin()}")

print(f"\n⚡ FACILITY VERBRUIK (kWh):")
print(f"  Totaal: {fac_aligned.sum():,.0f} kWh")
print(f"  Gemiddeld per uur: {fac_aligned.mean():.2f} kWh/uur")
print(f"  Piek: {fac_aligned.max():.2f} kWh/uur op {fac_aligned.idxmax()}")
print(f"  Minimum: {fac_aligned.min():.2f} kWh/uur op {fac_aligned.idxmin()}")

# Totaal en percentage
total_combined = ev_aligned.sum() + fac_aligned.sum()
ev_pct_total = (ev_aligned.sum() / total_combined * 100)

print(f"\n📈 GECOMBINEERD:")
print(f"  Totaal (EV + Facility): {total_combined:,.0f} kWh")
print(f"  EV aandeel: {ev_pct_total:.1f}%")
print(f"  Facility aandeel: {100-ev_pct_total:.1f}%")

print("\n💡 CONCLUSIE:")
if ev_pct_total > 20:
    print(f"  ✓ EV heeft GROTE invloed ({ev_pct_total:.1f}% van totaal verbruik)")
elif ev_pct_total > 10:
    print(f"  ◐ EV heeft MATIGE invloed ({ev_pct_total:.1f}% van totaal verbruik)")
else:
    print(f"  ✗ EV heeft BEPERKTE invloed ({ev_pct_total:.1f}% van totaal verbruik)")

print("\n" + "="*70)
```

```python
# === DETAIL: 9 SEPTEMBER 2025 - EV vs Facility (kWh/uur) ===
plot_date = pd.to_datetime('2025-09-09').date()

# Extract day data
try:
    ev_day = ev_aligned[ev_aligned.index.date == plot_date]
    fac_day = fac_aligned[fac_aligned.index.date == plot_date]
except NameError:
    # fall back to original series if aligned not present
    ev_day = ev_total_sep[ev_total_sep.index.date == plot_date]
    facility_kwh_hour = facility_kwh_hour if 'facility_kwh_hour' in globals() else facility_kwh_hour
    fac_day = facility_kwh_hour[facility_kwh_hour.index.date == plot_date]

if ev_day.empty and fac_day.empty:
    print(f"Geen data gevonden voor {plot_date}")
else:
    fig, ax = plt.subplots(figsize=(14,6))
    ax.plot(ev_day.index, ev_day.values, marker='o', linestyle='-', color='steelblue', label='EV (kWh/uur)')
    ax.plot(fac_day.index, fac_day.values, marker='o', linestyle='-', color='crimson', label='Facility (kWh/uur)')
    ax.set_title(f'9 September 2025 - EV vs Facility (kWh/uur)')
    ax.set_xlabel('Tijd')
    ax.set_ylabel('kWh/uur')
    ax.legend()
    ax.grid(alpha=0.3)
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

    # Dagtotalen
    ev_total_day = ev_day.sum()
    fac_total_day = fac_day.sum()
    combined = ev_total_day + fac_total_day
    ev_pct_day = (ev_total_day / combined * 100) if combined>0 else 0

    print('\n📅 9 September 2025 - Dagstatistieken')
    print(f'  EV totaal (kWh): {ev_total_day:,.2f}')
    print(f'  Facility totaal (kWh): {fac_total_day:,.2f}')
    print(f'  EV aandeel: {ev_pct_day:.1f}%')

    # Staafdiagram totalen
    fig, ax = plt.subplots(figsize=(6,4))
    ax.bar(['EV','Facility'], [ev_total_day, fac_total_day], color=['steelblue','crimson'], alpha=0.8)
    ax.set_ylabel('kWh/dag')
    ax.set_title('9 Sept 2025 - Totale kWh')
    for i, v in enumerate([ev_total_day, fac_total_day]):
        ax.text(i, v + max(ev_total_day, fac_total_day)*0.01, f'{v:,.1f}', ha='center')
    plt.tight_layout()
    plt.show()
```

```python
df_laadprofielen = pd.read_excel('data/Sessions (1).xlsx')

df_laadprofielen.head()
```

```python
# Maak kolom voor classificatie: visitor (geen token) vs werknemer (heeft token)
df_laadprofielen['User_Type'] = df_laadprofielen['Token'].apply(
    lambda x: 'werknemer' if pd.notna(x) else 'visitor'
)

df_laadprofielen.head()
```

```python
# === GRAFIEK: USER TYPE VERHOUDING ===

# Bereken aantallen
user_counts = df_laadprofielen['User_Type'].value_counts()
user_pcts = df_laadprofielen['User_Type'].value_counts(normalize=True) * 100

# Maak figuur met 2 subplots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

# === Plot 1: Pie Chart ===
colors_pie = ['#FF6B6B', '#4ECDC4']  # rood voor werknemers, turquoise voor visitors
wedges, texts, autotexts = ax1.pie(
    user_counts.values, 
    labels=user_counts.index, 
    autopct='%1.1f%%',
    startangle=90,
    colors=colors_pie,
    textprops={'fontsize': 12, 'fontweight': 'bold'},
    explode=(0.05, 0.05)
)

# Zet percentage waarden beter zichtbaar
for autotext in autotexts:
    autotext.set_color('white')
    autotext.set_fontsize(13)
    autotext.set_fontweight('bold')

ax1.set_title('User Type Verhouding\n(Pie Chart)', fontsize=14, fontweight='bold', pad=20)

# === Plot 2: Bar Chart met aantallen ===
bars = ax2.bar(
    ['👤 Werknemers\n(met token)', '🚗 Visitors\n(zonder token)'],
    user_counts.values,
    color=colors_pie,
    alpha=0.8,
    edgecolor='black',
    linewidth=1.5
)

# Voeg waarden op bars toe
for bar, count, pct in zip(bars, user_counts.values, user_pcts.values):
    height = bar.get_height()
    ax2.text(
        bar.get_x() + bar.get_width()/2., 
        height,
        f'{count:,}\n({pct:.1f}%)',
        ha='center', 
        va='bottom',
        fontsize=11,
        fontweight='bold'
    )

ax2.set_ylabel('Aantal Sessies', fontsize=12, fontweight='bold')
ax2.set_title('User Type Verdeling\n(Bar Chart)', fontsize=14, fontweight='bold', pad=20)
ax2.grid(True, alpha=0.3, axis='y')
ax2.set_ylim(0, max(user_counts.values) * 1.15)

plt.tight_layout()
plt.show()

```

```python
# === GROOTSTE LAADPAAL VERBRUIK PER LAADPAAL ===

# Groepeer per laadpaal en find maximum verbruik
max_kwh_per_station = df_laadprofielen.groupby('Charging station name')['kWh'].max().sort_values(ascending=False)

print("\n" + "="*70)
print("⚡ GROOTSTE LAADPAAL VERBRUIK PER LAADPAAL")
print("="*70)

for idx, (station, max_kwh) in enumerate(max_kwh_per_station.items(), 1):
    print(f"{idx:2d}. {station:25s} → {max_kwh:7.2f} kWh")

print("\n" + "="*70)

# Bereik statistieken per laadpaal
stats_per_station = df_laadprofielen.groupby('Charging station name')['kWh'].agg([
    ('Sessies', 'count'),
    ('Min [kWh]', 'min'),
    ('Max [kWh]', 'max'),
    ('Gem [kWh]', 'mean'),
    ('Totaal [kWh]', 'sum')
]).sort_values('Max [kWh]', ascending=False)

# === GRAFIEK 1: MAX VERBRUIK PER LAADPAAL (Horizontal Bar) ===
fig, ax = plt.subplots(figsize=(12, 8))

bars = ax.barh(
    range(len(max_kwh_per_station)),
    max_kwh_per_station.values,
    color='#FF6B6B',
    edgecolor='black',
    linewidth=1.5,
    alpha=0.8
)

for i, (bar, value) in enumerate(zip(bars, max_kwh_per_station.values)):
    ax.text(value, bar.get_y() + bar.get_height()/2, f' {value:.2f} kWh', 
            va='center', fontsize=11, fontweight='bold')

ax.set_yticks(range(len(max_kwh_per_station)))
ax.set_yticklabels(max_kwh_per_station.index, fontsize=11)
ax.set_xlabel('Maximaal Verbruik per Sessie [kWh]', fontsize=12, fontweight='bold')
ax.set_title('Grootste Laadpaal Verbruik per Laadpaal', fontsize=14, fontweight='bold', pad=20)
ax.grid(True, alpha=0.3, axis='x')

plt.tight_layout()
plt.show()

# === GRAFIEKEN 2-5: AANVULLENDE STATISTIEKEN (4x1 grid) ===
fig, axes = plt.subplots(2, 2, figsize=(16, 12))
fig.suptitle('Laadpaal Statistieken - Gedetailleerde Analyse', fontsize=16, fontweight='bold')

# Plot 1: Sessies per laadpaal
ax = axes[0, 0]
bars1 = ax.bar(range(len(stats_per_station)), stats_per_station['Sessies'].values, 
               color='#4ECDC4', edgecolor='black', linewidth=1.5, alpha=0.8)
ax.set_ylabel('Aantal Sessies', fontsize=12, fontweight='bold')
ax.set_title('Sessies per Laadpaal', fontsize=13, fontweight='bold')
ax.set_xticks(range(len(stats_per_station)))
ax.set_xticklabels(stats_per_station.index, rotation=45, ha='right', fontsize=10)
ax.grid(True, alpha=0.3, axis='y')

for bar, val in zip(bars1, stats_per_station['Sessies'].values):
    height = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2., height, f'{int(val)}',
            ha='center', va='bottom', fontsize=10, fontweight='bold')

# Plot 2: Totaal verbruik per laadpaal
ax = axes[0, 1]
bars2 = ax.bar(range(len(stats_per_station)), stats_per_station['Totaal [kWh]'].values, 
               color='#FF6B6B', edgecolor='black', linewidth=1.5, alpha=0.8)
ax.set_ylabel('Totaal Verbruik [kWh]', fontsize=12, fontweight='bold')
ax.set_title('Totaal Verbruik per Laadpaal', fontsize=13, fontweight='bold')
ax.set_xticks(range(len(stats_per_station)))
ax.set_xticklabels(stats_per_station.index, rotation=45, ha='right', fontsize=10)
ax.grid(True, alpha=0.3, axis='y')

for bar, val in zip(bars2, stats_per_station['Totaal [kWh]'].values):
    height = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2., height, f'{val:,.0f}',
            ha='center', va='bottom', fontsize=9, fontweight='bold')

# Plot 3: Gemiddeld verbruik per laadpaal
ax = axes[1, 0]
bars3 = ax.bar(range(len(stats_per_station)), stats_per_station['Gem [kWh]'].values, 
               color='#95E1D3', edgecolor='black', linewidth=1.5, alpha=0.8)
ax.set_ylabel('Gemiddeld Verbruik [kWh]', fontsize=12, fontweight='bold')
ax.set_title('Gemiddeld Verbruik per Laadpaal', fontsize=13, fontweight='bold')
ax.set_xticks(range(len(stats_per_station)))
ax.set_xticklabels(stats_per_station.index, rotation=45, ha='right', fontsize=10)
ax.grid(True, alpha=0.3, axis='y')

for bar, val in zip(bars3, stats_per_station['Gem [kWh]'].values):
    height = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2., height, f'{val:.2f}',
            ha='center', va='bottom', fontsize=10, fontweight='bold')

# Plot 4: Min vs Max verbruik per laadpaal
ax = axes[1, 1]
x_pos = range(len(stats_per_station))
width = 0.35

bars_min = ax.bar([i - width/2 for i in x_pos], stats_per_station['Min [kWh]'].values, 
                  width, label='Min', color='#A8E6CF', edgecolor='black', linewidth=1.5, alpha=0.8)
bars_max = ax.bar([i + width/2 for i in x_pos], stats_per_station['Max [kWh]'].values, 
                  width, label='Max', color='#FF8B94', edgecolor='black', linewidth=1.5, alpha=0.8)

ax.set_ylabel('Verbruik [kWh]', fontsize=12, fontweight='bold')
ax.set_title('Min vs Max Verbruik per Laadpaal', fontsize=13, fontweight='bold')
ax.set_xticks(x_pos)
ax.set_xticklabels(stats_per_station.index, rotation=45, ha='right', fontsize=10)
ax.legend(fontsize=11, loc='upper right')
ax.grid(True, alpha=0.3, axis='y')

plt.tight_layout()
plt.show()
```

```python
# === VERHOUDINGEN: VISITOR vs WERKNEMER (Laadduur & Verbruik) ===

# Voorbereiding: converteer Duration [h:mm] naar uren
def duration_to_hours(duration_str):
    """Converteer 'h:mm' format naar uren (float)"""
    try:
        if pd.isna(duration_str):
            return 0
        parts = str(duration_str).split(':')
        hours = int(parts[0])
        minutes = int(parts[1])
        return hours + minutes / 60
    except:
        return 0

df_laadprofielen['Duration_hours'] = df_laadprofielen['Duration [h:mm]'].apply(duration_to_hours)

print("\n" + "="*70)
print("👥 VISITOR vs WERKNEMER - LAADDUUR & VERBRUIK VERGELIJKING")
print("="*70)

for user_type in ['visitor', 'werknemer']:
    subset = df_laadprofielen[df_laadprofielen['User_Type'] == user_type]
    print(f"\n{user_type.upper()}:")
    print(f"  Sessies: {len(subset):,}")
    print(f"  Totaal verbruik: {subset['kWh'].sum():,.2f} kWh")
    print(f"  Gem. verbruik/sessie: {subset['kWh'].mean():.2f} kWh")
    print(f"  Totaal laadduur: {subset['Duration_hours'].sum():,.1f} uur")
    print(f"  Gem. laadduur/sessie: {subset['Duration_hours'].mean():.2f} uur ({subset['Duration_hours'].mean()*60:.0f} min)")
    if subset['Duration_hours'].sum() > 0:
        print(f"  Gem. vermogen (kW): {(subset['kWh'].sum() / subset['Duration_hours'].sum()):.2f}")

print("\n" + "="*70)

# === GRAFIEKEN ===
fig, axes = plt.subplots(2, 3, figsize=(18, 10))
fig.suptitle('Visitor vs Werknemer - Laadduur & Verbruik Analyse', fontsize=16, fontweight='bold')

# Voorbereiding kleuren
colors_usertype = {'visitor': '#4ECDC4', 'werknemer': '#FF6B6B'}

# === Plot 1: Aantal sessies ===
ax = axes[0, 0]
user_type_counts = df_laadprofielen['User_Type'].value_counts()
bars = ax.bar(user_type_counts.index, user_type_counts.values, 
              color=[colors_usertype[ut] for ut in user_type_counts.index],
              edgecolor='black', linewidth=2, alpha=0.8)
ax.set_ylabel('Aantal Sessies', fontsize=11, fontweight='bold')
ax.set_title('Sessies per User Type', fontsize=12, fontweight='bold')
ax.grid(True, alpha=0.3, axis='y')
for bar, val in zip(bars, user_type_counts.values):
    height = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2., height, f'{int(val):,}',
            ha='center', va='bottom', fontsize=11, fontweight='bold')

# === Plot 2: Totaal verbruik per user type ===
ax = axes[0, 1]
total_kwh = df_laadprofielen.groupby('User_Type')['kWh'].sum()
bars = ax.bar(total_kwh.index, total_kwh.values,
              color=[colors_usertype[ut] for ut in total_kwh.index],
              edgecolor='black', linewidth=2, alpha=0.8)
ax.set_ylabel('Totaal Verbruik [kWh]', fontsize=11, fontweight='bold')
ax.set_title('Totaal Verbruik per User Type', fontsize=12, fontweight='bold')
ax.grid(True, alpha=0.3, axis='y')
for bar, val in zip(bars, total_kwh.values):
    height = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2., height, f'{val:,.0f}',
            ha='center', va='bottom', fontsize=11, fontweight='bold')

# === Plot 3: Gemiddeld verbruik per sessie ===
ax = axes[0, 2]
avg_kwh = df_laadprofielen.groupby('User_Type')['kWh'].mean()
bars = ax.bar(avg_kwh.index, avg_kwh.values,
              color=[colors_usertype[ut] for ut in avg_kwh.index],
              edgecolor='black', linewidth=2, alpha=0.8)
ax.set_ylabel('Gemiddeld Verbruik [kWh]', fontsize=11, fontweight='bold')
ax.set_title('Gemiddeld Verbruik per Sessie', fontsize=12, fontweight='bold')
ax.grid(True, alpha=0.3, axis='y')
for bar, val in zip(bars, avg_kwh.values):
    height = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2., height, f'{val:.2f}',
            ha='center', va='bottom', fontsize=11, fontweight='bold')

# === Plot 4: Totaal laadduur ===
ax = axes[1, 0]
total_hours = df_laadprofielen.groupby('User_Type')['Duration_hours'].sum()
bars = ax.bar(total_hours.index, total_hours.values,
              color=[colors_usertype[ut] for ut in total_hours.index],
              edgecolor='black', linewidth=2, alpha=0.8)
ax.set_ylabel('Totaal Laadduur [uur]', fontsize=11, fontweight='bold')
ax.set_title('Totaal Laadduur per User Type', fontsize=12, fontweight='bold')
ax.grid(True, alpha=0.3, axis='y')
for bar, val in zip(bars, total_hours.values):
    height = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2., height, f'{val:,.0f}h',
            ha='center', va='bottom', fontsize=11, fontweight='bold')

# === Plot 5: Gemiddelde laadduur per sessie ===
ax = axes[1, 1]
avg_hours = df_laadprofielen.groupby('User_Type')['Duration_hours'].mean()
bars = ax.bar(avg_hours.index, avg_hours.values,
              color=[colors_usertype[ut] for ut in avg_hours.index],
              edgecolor='black', linewidth=2, alpha=0.8)
ax.set_ylabel('Gemiddelde Laadduur [uur]', fontsize=11, fontweight='bold')
ax.set_title('Gemiddelde Laadduur per Sessie', fontsize=12, fontweight='bold')
ax.grid(True, alpha=0.3, axis='y')
for bar, val in zip(bars, avg_hours.values):
    height = bar.get_height()
    mins = val * 60
    ax.text(bar.get_x() + bar.get_width()/2., height, f'{val:.2f}h\n({mins:.0f}m)',
            ha='center', va='bottom', fontsize=10, fontweight='bold')

# === Plot 6: Laad vermogen (kWh/uur) vergelijking ===
ax = axes[1, 2]
visitor_kwh = df_laadprofielen[df_laadprofielen['User_Type'] == 'visitor']
werknemer_kwh = df_laadprofielen[df_laadprofielen['User_Type'] == 'werknemer']

visitor_power = (visitor_kwh['kWh'].sum() / visitor_kwh['Duration_hours'].sum()) if visitor_kwh['Duration_hours'].sum() > 0 else 0
werknemer_power = (werknemer_kwh['kWh'].sum() / werknemer_kwh['Duration_hours'].sum()) if werknemer_kwh['Duration_hours'].sum() > 0 else 0

bars = ax.bar(['visitor', 'werknemer'], [visitor_power, werknemer_power],
              color=[colors_usertype['visitor'], colors_usertype['werknemer']],
              edgecolor='black', linewidth=2, alpha=0.8)
ax.set_ylabel('Gemiddeld Vermogen [kW]', fontsize=11, fontweight='bold')
ax.set_title('Laadvermogen per User Type\n(totaal kWh / totaal uren)', fontsize=12, fontweight='bold')
ax.grid(True, alpha=0.3, axis='y')
for bar, val in zip(bars, [visitor_power, werknemer_power]):
    height = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2., height, f'{val:.2f} kW',
            ha='center', va='bottom', fontsize=11, fontweight='bold')

plt.tight_layout()
plt.show()

# === TABEL SAMENVATTING ===
print("\n" + "="*70)
print("📊 GEDETAILLEERDE STATISTIEKEN")
print("="*70)

summary_data = {
    'Metric': ['Sessies', 'Totaal kWh', 'Gem. kWh/sessie', 'Totaal uren', 'Gem. uren/sessie', 'Gem. minuten/sessie', 'Gem. vermogen (kW)'],
    'Visitor': [
        len(visitor_kwh),
        f'{visitor_kwh["kWh"].sum():,.0f}',
        f'{visitor_kwh["kWh"].mean():.2f}',
        f'{visitor_kwh["Duration_hours"].sum():,.0f}',
        f'{visitor_kwh["Duration_hours"].mean():.2f}',
        f'{visitor_kwh["Duration_hours"].mean() * 60:.0f}',
        f'{visitor_power:.2f}'
    ],
    'Werknemer': [
        len(werknemer_kwh),
        f'{werknemer_kwh["kWh"].sum():,.0f}',
        f'{werknemer_kwh["kWh"].mean():.2f}',
        f'{werknemer_kwh["Duration_hours"].sum():,.0f}',
        f'{werknemer_kwh["Duration_hours"].mean():.2f}',
        f'{werknemer_kwh["Duration_hours"].mean() * 60:.0f}',
        f'{werknemer_power:.2f}'
    ]
}

summary_df = pd.DataFrame(summary_data)
print("\n" + summary_df.to_string(index=False))

print("\n" + "="*70)
```
